```python
Crime.columns = Crime.columns.str.lower().str.strip()
Income.columns = Income.columns.str.lower().str.strip()
CrimeReport.columns = CrimeReport.columns.str.lower().str.strip()
```

```python
Income.rename(
    columns={"financial_year": "year"},
    inplace=True
)
```

```python
Income["year"] = Income["year"].dt.year
```

```python
Income.head()
```

```python
Crime_clean = Crime.copy()

Income_clean = Income.copy()

CrimeReport_clean = CrimeReport.copy()
```

```python
Crime_clean.drop("source", axis=1, inplace=True)
Income_clean.drop("source", axis=1, inplace=True)
CrimeReport_clean.drop("source", axis=1, inplace=True)
```

```python
crime_cols = [
    "year",
    "burglary_cases",
    "theft_cases",
    "robbery_cases",
    "murder_cases"
]

Crime_analysis = Crime_clean[crime_cols].copy()
```

```python
Crime_analysis
```

```python
Master = pd.merge(
    Crime_analysis,
    Income_clean,
    on="year",
    how="inner"
)
```

```python
Master
```

```python
Master["year"].is_unique
```

```python
Master["per_capita_income"] = Master["delhi_per_capita_income_constant_rs"]
```

```python
Master
```

```python
Master["burglary_pct_change"] = Master["burglary_cases"].pct_change() * 100
```

```python
Master["income_pct_change"] = Master["per_capita_income"].pct_change() * 100
```

```python
Master
```

```python
Master["theft_pct_change"] = Master["theft_cases"].pct_change() * 100

Master["robbery_pct_change"] = Master["robbery_cases"].pct_change() * 100

Master["murder_pct_change"] = Master["murder_cases"].pct_change() * 100
```

```python
Master
```

```python
correlation = Master["burglary_cases"].corr(
    Master["per_capita_income"]
)

print(correlation)
```

```python
correlation_change = Master["burglary_pct_change"].corr(

    Master["income_pct_change"]

)

print(correlation_change)
```

```python
Unemployment = pd.DataFrame({

    "year": [2018, 2019, 2020, 2021],

    "unemployment_rate": [10.4, 8.6, 6.3, 5.3]

})
```

```python
Master = pd.merge(
    Master,
    Unemployment,
    on="year",
    how="left"
)
```

```python
correlation_burglary_unemployment = Master["burglary_cases"].corr(

    Master["unemployment_rate"]

)

print(correlation_burglary_unemployment)
```

```python
Population = pd.DataFrame({

    "year": [2018, 2019, 2020, 2021],

    "population": [

        19560000,

        19930000,

        20310000,

        20700000

    ]

})
```

```python
Master = pd.merge(

    Master,

    Population,

    on="year",

    how="left"

)
```

```python
Master["burglary_rate"] = (
    Master["burglary_cases"] / Master["population"]
) * 100000
```

```python
Master[["year", "burglary_cases", "population", "burglary_rate"]]
```

```python
correlation_rate = Master["burglary_rate"].corr(
    Master["unemployment_rate"]
)

print(correlation_rate)
```

```python
import matplotlib.pyplot as plt

import seaborn as sns

plt.figure(figsize=(8, 5))

sns.regplot(

    data=Master,

    x="unemployment_rate",

    y="burglary_rate"

)

plt.xlabel("Unemployment Rate (%)")

plt.ylabel("Burglary Rate (per 100,000)")

plt.title("Burglary Rate vs Unemployment in Delhi (2018–2021)")

plt.show()
```
